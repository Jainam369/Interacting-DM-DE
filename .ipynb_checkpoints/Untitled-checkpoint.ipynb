```python
from classy import Class
import numpy as np

def run_point(H0, omega_b, ns, As, tau, Omega_c, beta, log_mphi_H0, delta_i):
    cosmo = Class()
    cosmo.set({
        'h': H0/100,
        'omega_b': omega_b,
        'n_s': ns,
        'A_s': As,
        'tau_reio': tau,
        'omega_idm': Omega_c * (H0/100)**2,
        'scf_potential': 'axion',
        'n_axion': 1,
        'm_axion': 10**log_mphi_H0,
        'scf_parameters': f'{np.pi - delta_i}, 0',
        'attractor_ic_scf': 'no',
        'do_shooting_scf': 'no',
        'beta_axion': beta,
        'DMDE_interaction': 0,
        'loop_over_background_for_closure_relation': 'yes',
        'precision_loop_over_background' : 1e-2,
        'evolver' : 0,
        'Omega_scf_shoot_fa': 1 - 0.0493 - Omega_c - 9.2e-5,  # rough target; refine below
    })
    cosmo.compute()
    omega_scf = cosmo.get_current_derived_parameters(['Omega_scf'])  # check actual getter name
    cosmo.struct_cleanup()
    cosmo.empty()
    return omega_scf

# Test with a Table II-like point
result = run_point(67.96, 0.02237, 0.9649, 2.1e-9, 0.0544, 0.3046, 0.077, -0.346, 0.839)
print(result)
```
